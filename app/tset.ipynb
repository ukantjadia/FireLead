# FireLead Project Demo

This notebook demonstrates the main features of the FireLead project, showcasing both the Firecrawl extraction and enhanced extraction capabilities.

## Setup and Imports

```python
import os
import pandas as pd
from dotenv import load_dotenv
from extractor import FirecrawlDataExtractor
from data_saver import DataSaver
import json

# Load environment variables
load_dotenv()

# Get API key from environment variable
FIRECRAWL_API_KEY = os.getenv('FIRECRAWL_API_KEY')
if not FIRECRAWL_API_KEY:
    raise ValueError("FIRECRAWL_API_KEY environment variable is not set")

# Initialize extractor and data saver
extractor = FirecrawlDataExtractor(api_key=FIRECRAWL_API_KEY)
data_saver = DataSaver('data')
```

## 1. Basic Firecrawl Extraction

This section demonstrates the basic Firecrawl extraction functionality, which allows you to extract leads from various sources using a natural language prompt.

```python
# Define the prompt and fields to extract
prompt = "Find companies in the AI/ML space that have raised funding in the last 6 months"

# Define which fields we want to extract
selected_fields = {
    "company_name": True,
    "funding": True,
    "industry": True,
    "description": True,
    "website_url": True,
    "email": True
}

# Define URL patterns to search
url_patterns = [
    "https://firecrawl.dev/*",
    "https://tracxn.com/*",
    "https://www.cbinsights.com/*",
    "https://pitchbook.com/*",
    "https://www.crunchbase.com/*"
]

print("Starting Firecrawl extraction...")
results = extractor.extract_leads(url_patterns, prompt, selected_fields, enable_web_search=True)

if results and 'data' in results and 'leads' in results['data']:
    # Convert to DataFrame
    df = pd.DataFrame(results['data']['leads'])
    print(f"\nFound {len(df)} leads")
    print("\nSample of extracted data:")
    display(df.head())

    # Save the results
    saved_file = data_saver.save(df)
    print(f"\nData saved to: {saved_file.name}")
else:
    print("No leads found in the extraction results")
```

## 2. Enhanced Extraction

This section demonstrates the enhanced extraction feature, which allows for more customized field extraction with specific data types.

```python
# Define a more specific prompt
enhanced_prompt = "Find out the founder and CEO of the company, the year it was founded, and the number of employees. Also, check if they have a free trial available and what their target market is. Provide key features of the product as well."
url_patterns = ["https://unimans.com/", "https://www.brainguru.in/", "https://www.saasworthy.com/"]
# Define specific fields with their types
specific_fields = {
    "company_name": "str",
    "product_name": "str",
    "founding_year": "str",
    "ceo_name": "str",
    "number_of_employees": "number",
    "free_trial_available": "bool",
    "target_market": "str",
}

print("Starting enhanced extraction...")
enhanced_results = extractor.extract_leads_enhanced(
    url_patterns,
    enhanced_prompt,
    specific_fields,
    enable_web_search=True
)
print(enhanced_results)

if enhanced_results and 'data' in enhanced_results and 'leads' in enhanced_results['data']:
    # Convert to DataFrame
    enhanced_df = pd.DataFrame(enhanced_results['data']['leads'])
    print(f"\nFound {len(enhanced_df)} leads")
    print("\nSample of enhanced extraction data:")
    display(enhanced_df.head())

    # Save the results
    saved_file = data_saver.save(enhanced_df)
    print(f"\nData saved to: {saved_file.name}")
else:
    print("No leads found in the enhanced extraction results")
```
